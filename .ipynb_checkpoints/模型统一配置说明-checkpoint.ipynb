## 硅基流动

```python
import os
from dotenv import load_dotenv, find_dotenv
# 读取本地/项目的环境变量。
# find_dotenv() 寻找并定位 .env 文件的路径
# load_dotenv() 读取该 .env 文件，并将其中的环境变量加载到当前的运行环境中  
# _ = load_dotenv("../.env") # 或者手动指定
_ = load_dotenv(find_dotenv())

# ── 1. 大语言模型配置 ─────────────────────────────
siliconflow_base_url = os.getenv("SILICONFLOW_BASE_URL")
siliconflow_api_key = os.getenv("SILICONFLOW_API_KEY")

from openai import OpenAI

client = OpenAI(api_key=siliconflow_api_key, base_url=siliconflow_base_url)

response = client.chat.completions.create(
    model="Qwen/Qwen3-8B",
    messages=[
        {"role": "system", "content": "你是一个天气助手"},
        {"role": "user", "content": "你是谁？"}
    ],
    stream=False
)

print(response.choices[0].message.content)
```

## DeepSeek

```python
import os
from dotenv import load_dotenv, find_dotenv
from openai import OpenAI

load_dotenv(find_dotenv())

client = OpenAI(
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    api_key=os.getenv("DEEPSEEK_API_KEY")
)

response = client.chat.completions.create(
    model="deepseek-v4-flash",
    messages=[
        {"role": "system", "content": "你是一个天气助手"},
        {"role": "user", "content": "你是谁？"}
    ],
    stream=False
)

print(response.choices[0].message.content)
```

```python
import os
import httpx
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model


# 1. 加载环境变量
load_dotenv(find_dotenv())

# 2. 初始化大语言模型
my_llm = init_chat_model(
    model=os.getenv("DEEPSEEK_MODEL_ID"),
    model_provider="openai",
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    http_client=httpx.Client(trust_env=False),
    temperature=0.3,
    max_tokens=3000,
)

# 3. 调用模型，只输出 content
response = my_llm.invoke("你是谁？")
print(response.content)
```

```python
import os
from dotenv import load_dotenv, find_dotenv
from openai import OpenAI

load_dotenv(find_dotenv())

client = OpenAI(
    base_url=os.getenv("vLLM_QWEN8B_BASE_URL"),
    api_key=os.getenv("vLLM_QWEN8B_API_KEY"),
)

response = client.chat.completions.create(
    model=os.getenv("vLLM_QWEN8B_MODEL_ID"),
    messages=[
        {"role": "system", "content": "你是一个天气助手"},
        {"role": "user", "content": "你是谁？"}
    ],
    stream=False
)

print(response.choices[0].message.content)
```
